```python
import pandas as pd

# 1. Carregar o arquivo CSV original
df = pd.read_csv('censo_eua.csv')

# 2. Dicionários de tradução para os valores internos de cada coluna categórica
trad_workclass = {
    'State-gov': 'Gov-Estadual',
    'Self-emp-not-inc': 'Autonomo_sem_inc',
    'Private': 'Privado',
    'Federal-gov': 'Gov-Federal',
    'Local-gov': 'Gov-Local',
    'Self-emp-inc': 'Autonomo_com_inc',
    'Without-pay': 'Sem_remuneracao',
    'Never-worked': 'Nunca_trabalhou',
}

trad_education = {
    'Bachelors': 'Superior_Bacharelado',
    'HS-grad': 'Ensino_Medio',
    '11th': '11_Ano',
    'Masters': 'Mestrado',
    '9th': '9_Ano',
    'Some-college': 'Superior_Incompleto',
    'Assoc-acdm': 'Tecnologo_Academico',
    'Assoc-voc': 'Tecnologo_Profissional',
    '7th-8th': '7_8_Ano',
    'Doctorate': 'Doutorado',
    'Prof-school': 'Curso_Profissionalizante',
    '5th-6th': '5_6_Ano',
    '10th': '10_Ano',
    '1st-4th': '1_4_Ano',
    'Preschool': 'Pre_escola',
    '12th': '12_Ano',
}

trad_marital_status = {
    'Never-married': 'Solteiro(a)',
    'Married-civ-spouse': 'Casado(a)',
    'Divorced': 'Divorciado(a)',
    'Married-spouse-absent': 'Casado(a)_conjuge_ausente',
    'Separated': 'Separado(a)',
    'Married-AF-spouse': 'Casado(a)_forcas_armadas',
    'Widowed': 'Viuvo(a)',
}

trad_occupation = {
    'Adm-clerical': 'Administrativo/Secretariado',
    'Exec-managerial': 'Executivo/Gerencial',
    'Handlers-cleaners': 'Operario/Limpeza',
    'Prof-specialty': 'Especialista_Profissional',
    'Other-service': 'Outros_Servicos',
    'Sales': 'Vendas',
    'Craft-repair': 'Artesao/Manutencao',
    'Transport-moving': 'Transporte/Motorista',
    'Farming-fishing': 'Agricultura/Pesca',
    'Machine-op-inspct': 'Operador_de_Maquinas/Inspecao',
    'Tech-support': 'Suporte_Tecnico',
    'Protective-serv': 'Servicos_de_Protecao',
    'Armed-Forces': 'Forcas_Armadas',
    'Priv-house-serv': 'Servicos_Domesticos',
}

trad_relationship = {
    'Not-in-family': 'Fora_da_familia',
    'Husband': 'Marido',
    'Wife': 'Esposa',
    'Own-child': 'Filho(a)',
    'Unmarried': 'Nao_casado(a)',
    'Other-relative': 'Outro_parente',
}

trad_race = {
    'White': 'Branca',
    'Black': 'Preta',
    'Asian-Pac-Islander': 'Asiatico/Ilhas_Pacifico',
    'Amer-Indian-Eskimo': 'Indigena_Americano/Eskimo',
    'Other': 'Outra',
}

trad_sex = {'Male': 'Masculino', 'Female': 'Feminino'}

trad_native_country = {
    'United-States': 'Estados-Unidos',
    'Mexico': 'Mexico',
    'Philippines': 'Filipinas',
    'Germany': 'Alemanha',
    'Puerto-Rico': 'Porto-Rico',
    'Canada': 'Canada',
    'El-Salvador': 'El-Salvador',
    'India': 'India',
    'Cuba': 'Cuba',
    'England': 'Inglaterra',
    'Jamaica': 'Jamaica',
    'South': 'Sul',
    'China': 'China',
    'Italy': 'Italia',
    'Dominican-Republic': 'Republica-Dominicana',
    'Vietnam': 'Vietna',
    'Guatemala': 'Guatemala',
    'Japan': 'Japao',
    'Poland': 'Polonia',
    'Columbia': 'Columbia',
    'Taiwan': 'Taiwan',
    'Haiti': 'Haiti',
    'Portugal': 'Portugal',
    'Iran': 'Ira',
    'Nicaragua': 'Nicaragua',
    'Peru': 'Peru',
    'Greece': 'Grecia',
    'France': 'Franca',
    'Ecuador': 'Equador',
    'Ireland': 'Irlanda',
    'Hong': 'Hong-Kong',
    'Cambodia': 'Camboja',
    'Trinadad&Tobago': 'Trinidad-Tobago',
    'Thailand': 'Tailandia',
    'Laos': 'Laos',
    'Yugoslavia': 'Iugoslávia',
    'Outlying-US(Guam-USVI-etc)': 'Outros_EUA',
    'Hungary': 'Hungria',
    'Honduras': 'Honduras',
    'Scotland': 'Escocia',
    'Holand-Netherlands': 'Holanda',
}

# 3. Aplicar as traduções nos dados (substituindo os valores)
df['workclass'] = df['workclass'].map(trad_workclass)
df['education'] = df['education'].map(trad_education)
df['marital-status'] = df['marital-status'].map(trad_marital_status)
df['occupation'] = df['occupation'].map(trad_occupation)
df['relationship'] = df['relationship'].map(trad_relationship)
df['race'] = df['race'].map(trad_race)
df['sex'] = df['sex'].map(trad_sex)
df['native-country'] = df['native-country'].map(trad_native_country)

# Criando também a coluna 'target' numérica (0 e 1) baseada na renda
df['target'] = df['income'].map({'<=50K': 0, '>50K': 1})

# 4. Dicionário de tradução dos nomes das colunas
traducao_colunas = {
    'age': 'idade',
    'workclass': 'classe_trabalho',
    'fnlwgt': 'peso_final',
    'education': 'escolaridade',
    'education-num': 'anos_estudo',
    'marital-status': 'estado_civil',
    'occupation': 'ocupacao',
    'relationship': 'relacionamento',
    'race': 'raca',
    'sex': 'sexo',
    'capital-gain': 'ganho_capital',
    'capital-loss': 'perda_capital',
    'hours-per-week': 'horas_por_semana',
    'native-country': 'pais_natal',
    'income': 'renda',
}

# 5. Renomear as colunas
df1 = df.rename(columns=traducao_colunas)

# 6. Visualizar o resultado final
print(df1.head(3))

# (Opcional) Salvar em um novo arquivo CSV traduzido
# df_totalmente_traduzido.to_csv('censo_eua_totalmente_traduzido.csv', index=False)
```

```python
#  Verificar valores nulos por coluna (quantidade e porcentagem)
print('=== VALORES NULOS ===')
nulos = df1.isnull().sum()
nulos_percentual = (df1.isnull().mean() * 100).round(2)

df_nulos = pd.DataFrame(
    {'Qtd_Nulos': nulos, 'Porcentagem_%': nulos_percentual}
)
# Mostra apenas colunas que possuem valores nulos
print(df_nulos[df_nulos['Qtd_Nulos'] > 0])

# 3. Verificar linhas duplicadas
duplicadas = df1.duplicated().sum()
print(f'\nTotal de linhas duplicadas: {duplicadas}')
```

```python
# Filtra o DataFrame para mostrar todas as linhas que possuem duplicatas
linhas_duplicadas = df1[df1.duplicated(keep=False)]

# Ordena pelo índice (ou por colunas) para facilitar a comparação visual lado a lado
linhas_duplicadas = linhas_duplicadas.sort_values(by=list(df1.columns))

print(linhas_duplicadas)
```
